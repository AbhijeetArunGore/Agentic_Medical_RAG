# Agentic Medical RAG — Interactive Jupyter Notebook Walkthrough

This notebook demonstrates step-by-step how the **Agentic Medical RAG** system operates using stateful LangGraph multi-agent orchestration and ChromaDB vector retrieval.

```python
import uuid
from app.agents.orchestrator import create_medical_graph
from app.rag.vector_store import VectorStoreManager
```

## Step 1: Initialize Knowledge Base

```python
vm = VectorStoreManager()
count = vm.seed_initial_knowledge_base()
print(f"Indexed {count} clinical guidelines into ChromaDB.")
```

## Step 2: Execute Multi-Agent Stateful Workflow

```python
app = create_medical_graph()
initial_state = {
    "patient_id": "PAT-2026-X88",
    "patient_history": "",
    "clinical_trends": [],
    "query": "Treatment options for rising creatinine",
    "retrieved_docs": [],
    "draft_briefing": "",
    "reasoning": "Initializing state",
    "verified": False,
    "contradictions": [],
    "final_report": "",
    "citations": []
}

config = {"configurable": {"thread_id": str(uuid.uuid4())}}
for event in app.stream(initial_state, config):
    for node, update in event.items():
        print(f"Node: {node.upper()}")
        if 'reasoning' in update:
            print(f"Reasoning: {update['reasoning']}\n")
```

## Step 3: Inspect Final Report

```python
final_state = app.get_state(config).values
print("Final Report:")
print(final_state.get("final_report", "SAFETY HALT FLAGGED"))
```
